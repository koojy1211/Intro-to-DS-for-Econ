```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0.0, 2.0*np.pi, 20)
fx = np.cos(x)
gx = np.sin(x)
```

```python
y = "Hello World, this is an infinite series: "
z = r"$\cos(x)$"
```

```python
plt.figure()
plt.title(y+z)

plt.plot(x, fx, label=z)
plt.plot(x, gx, label=r'$\sin(x)$')

plt.legend()

plt.show()
```
