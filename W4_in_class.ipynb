### Extending the List

```python
y = ['a', 'b', 'c']
z = [1, 2, 3]
```

```python
y + z
```

```python
y.append(z)
y
```

```python
y = ['a', 'b', 'c']
z = [1, 2, 3]

y.extend(z)
y
```

### Enumerate

```python
foo = ('good', 'luck')
enumfol = list(enumerate(foo))
enumfol
```

```python
enumfod = dict(enumerate(foo))
enumfod
```

```python
print(enumfol[0])
print(type(enumfol[0]))
print('\n')
print(enumfol[0][0])
print(type(enumfol[0][0]))
```
